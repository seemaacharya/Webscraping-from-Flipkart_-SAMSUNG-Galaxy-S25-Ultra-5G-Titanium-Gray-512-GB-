```python
import requests
```

```python
from bs4 import BeautifulSoup
```

```python
url = "https://www.flipkart.com/samsung-galaxy-s25-ultra-5g-titanium-gray-512-gb/product-reviews/itm4edf9ee3c6c2d?pid=MOBH8K8U2FZXCP2R&lid=LSTMOBH8K8U2FZXCP2R0C4UGD&marketplace=FLIPKART"
```

```python
headers = {"User-Agent":"Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/135.0.0.0 Safari/537.36"}
```

```python
response = requests.get(url, headers=headers)
```

```python
print(response.status_code)
```

```python
soup = BeautifulSoup(response.content,'html.parser')
```

```python
print(soup.prettify())
```

```python
#1)To extract the names
names = soup.find_all("p", class_="_2NsDsF AwS1CA")
names
```

```python
for name in names:
    print(name.text)
    print("")
    
```

```python
# to extract the names
# To extract the review_title
# To extract reviews
#to extract ratings
#to extract location
#to extract year/month
```

```python
#2) To extract the review_title
titles = soup.find_all("p", class_="z9E0IG")
```

```python
print(titles)
```

```python
for title in titles:
    print(title.text.strip())
    print("")
```

```python
#3) To extract reviews

reviews=soup.find_all("div", class_="ZmyHeo")
```

```python
print(reviews)
```

```python
for review in reviews:
    print(review.text.replace("READ MORE",""))
    print("")
    
#Here i was getting "READ MORE" after every reviews, hence removed.    
```

```python
#4) to extract ratings

ratings = soup.find_all("div", class_="XQDdHH Ga3i8K")
```

```python
print(ratings)
```

```python
for rating in ratings:
    print(rating.text)
    print("")
```

```python
#5) to extract location
locations = soup.find_all("p", class_="MztJPv")
```

```python
print(locations)
```

```python
for location in locations:
    print(location.text.replace("Certified Buyer,", "").strip())
    print("")
# It is normal to see more than one element(location,certified buyer etc..) under one class(class_="MztJPv").    
#Here I was getting "Certified Buyer," on left side    
```

```python
#6) to extract customer_names and when it was reviewed
reviewed_date = soup.find_all("p", class_="_2NsDsF")
# I found customer name and reviewed_date listed under one class (class_="_2NsDsF").
```

```python
print(reviewed_date)
```

```python
# Find all 'p' tags with class '_2NsDsF'
all_dates_and_names = soup.find_all("p", class_="_2NsDsF")

# Initialize empty lists for names and review dates
customer_names = []
reviewed_dates = []

# Loop through the elements and separate names and review dates
for item in all_dates_and_names:
    text = item.text.strip()
    if "ago" in text:  # Check if it's a review date
        reviewed_dates.append(text)
    else:  # Otherwise, it must be a customer name
        customer_names.append(text)


# Print the lists to verify
print("names:", customer_names)
print("Reviewed Dates:", reviewed_dates)
```

```python
# Print the lists to verify

print(reviewed_dates)
```

```python
# Create a DataFrame
import pandas as pd
```

```python
print(len(customer_names))
print(len(titles))
print(len(reviews))
print(len(locations))
print(len(reviewed_dates))
```

```python
df = pd.DataFrame({"Name":customer_names,
                   "review_titles":titles,
                   "review_description":reviews,
                   "loc":locations,
                   "date":reviewed_dates
                  })
```

```python
#save to csv file
df.to_csv("Flipkart_reviews1.csv", index=False, encoding='utf-8')
```
